```python
import pandas as pd
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import Dense, Dropout, BatchNormalization
import tensorflow as tf
```

```python
df = pd.read_csv('https://drive.google.com/uc?id=1roJ83AbgzDcvRr0Gwud0BmdUQx-oSG-w')
df.head()
```

```python
df = df.drop(columns='Id')
df.head()
```

```python
category = pd.get_dummies(df.Species, dtype=int)
category
```

```python
new_df = pd.concat([df, category], axis=1)
new_df = new_df.drop(columns='Species')
new_df
```

```python
dataset = new_df.values
dataset
```

```python
X = dataset[:,0:4]
y = dataset[:,4:7]

min_max_scaler = preprocessing.MinMaxScaler()
X_scale = min_max_scaler.fit_transform(X)
X_scale

X_train, X_test, Y_train, Y_test = train_test_split(X_scale, y, test_size=0.3)

model = Sequential([
    Dense(64, activation='relu', input_shape=(4,)),
    Dense(64, activation='relu'),
    Dense(3, activation='softmax')
])

model.compile(optimizer='Adam', loss='categorical_crossentropy', metrics=['accuracy'])

hist = model.fit(X_train, Y_train, epochs=100)
```

```python
model.evaluate(X_test, Y_test, batch_size=1)
```

```python
class myCallback(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs={}):
        if(logs.get('accuracy')>0.9):
            print("\n Accuracy has reached >90%")
            self.model_stop_training = True
callbacks = myCallback()

hist = model.fit(X_train, Y_train, epochs=100, callbacks=[callbacks])
```
